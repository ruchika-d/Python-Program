```python

# Student Marks, Percentage and Grade

name = input("Enter student name: ")
n = int(input("Enter number of subjects: "))

total = 0

for i in range(1, n + 1):
    mark = float(input(f"Enter mark for Subject {i}: "))
    total += mark

percentage = total / n

if percentage >= 90:
    grade = "A+"
    performance = "Outstanding"
elif percentage >= 80:
    grade = "A"
    performance = "Excellent"
elif percentage >= 70:
    grade = "B"
    performance = "Very Good"
elif percentage >= 60:
    grade = "C"
    performance = "Good"
elif percentage >= 50:
    grade = "D"
    performance = "Average"
else:
    grade = "F"
    performance = "Needs Improvement"

print("\n--- Academic Performance ---")
print("Student Name :", name)
print("Total Marks :", total)
print("Percentage :", round(percentage, 2), "%")
print("Grade :", grade)
print("Performance :", performance)

# Display a simple motivational message
if percentage >= 90:
    print("Excellent work! Keep it up!")
elif percentage >= 50:
    print("Good effort! Keep improving!")
else:
    print("Keep practising. You can do better!")
```

```python

# Electricity Billing System

def calculate_bill(units):
    bill = 0

    if units <= 100:
        bill = units * 1.50
    elif units <= 200:
        bill = 100 * 1.50 + (units - 100) * 2.50
    elif units <= 500:
        bill = 100 * 1.50 + 100 * 2.50 + (units - 200) * 4.00
    else:
        bill = 100 * 1.50 + 100 * 2.50 + 300 * 4.00 + (units - 500) * 6.00

    return bill


n = int(input("Enter number of customers: "))

for i in range(1, n + 1):
    print(f"\nCustomer {i}")

    name = input("Enter customer name: ")
    units = float(input("Enter units consumed: "))

    bill = calculate_bill(units)

    print("\n--- Electricity Bill ---")
    print("Customer Name :", name)
    print("Units Consumed :", units)
    print("Bill Amount : Rs.", round(bill, 2))

    # Energy-saving tips
    print("\n--- Energy Saving Tips ---")

    if units <= 100:
        print("Your electricity usage is low.")
        print("Continue using energy-efficient appliances!")

    elif units <= 300:
        print("Switch off unused lights and fans.")
        print("Use electricity wisely.")

    else:
        print("Your electricity usage is high.")
        print("Reduce the use of ACs and heaters.")
```

```python
# Menu-Driven Student Information System
students = {}
def add_student():
  roll = input("Enter Roll No: ")
  name = input("Enter Name: ")
  dept = input("Enter Department: ")
  students[roll] = {"Name": name, "Department": dept}
  print("Student added successfully.")
def view_students():
  if not students:
    print("No student records found.")
    return
  print("\n--- Student Records ---")
  for roll, data in students.items():
    print("Roll No:", roll,"| Name:", data["Name"],"| Department:", data["Department"])
def search_student():
  roll = input("Enter Roll No to search: ")
  if roll in students:
    print(students[roll])
  else:
    print("Student not found.")
def delete_student():
  roll = input("Enter Roll No to delete: ")
  if roll in students:
    del students[roll]
    print("Student deleted successfully.")
  else:
    print("Student not found.")
while True:
  print("\n1. Add Student")
  print("2. View Students")
  print("3. Search Student")
  print("4. Delete Student")
  print("5. Exit")
  choice = input("Enter your choice: ")
  if choice == "1":
    add_student()
  elif choice == "2":
    view_students()
  elif choice == "3":
    search_student()
  elif choice == "4":
    delete_student()
  elif choice == "5":
    print("Exiting...")
    break
  else:
    print("Invalid choice.")
```
